## 13. Baseline Audience Overview

The baseline provides a starting point for comparing audience size across platforms before weekly growth is measured.

```python
audience_baseline = baseline[
    baseline["Metric"].isin([
        "Monthly listeners",
        "Followers",
        "Subscribers"
    ])
].copy()

audience_baseline
```

### Baseline Audience by Platform

The chart below provides an initial overview of the recorded audience metrics. Platform figures should be interpreted separately because each platform measures audience size differently.

```python
plt.figure(figsize=(10, 6))

plt.bar(
    audience_baseline["Platform"] + " - " + audience_baseline["Metric"],
    audience_baseline["Baseline Value"]
)

plt.xticks(rotation=45, ha="right")
plt.ylabel("Baseline Value")
plt.title("@WAR Baseline Audience Metrics")
plt.tight_layout()
plt.show()
```
